```python
import tensorflow as tf
import matplotlib.pyplot as plt

# WE will load thedataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.fashion_mnist.load_data()
print(x_train.shape)
# Min-max Normalization
x_train = x_train/255.0
x_test = x_test/255.0
print(x_train.shape)
print(y_train.shape)
plt.imshow(x_train[0])
```

This ensures all pixel values are floats between 0 and 1 — important for training neural networks, especially CNNs, since it helps them converge faster and more reliably.

```python
x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)
```

🎯 Why Reshaping Is Necessary
CNNs require input in this format:

(batch_size,

height,

width,

channels)
(batch_size, height, width, channels)
So if you load an image dataset as just a 2D array or grayscale image (e.g., shape: (28, 28)), it must be reshaped to:

(
28
,
28
,
1
)
(28,28,1)
Or for a batch of 100 grayscale images:

(
100
,
28
,
28
,
1
)
(100,28,28,1)

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

datagen = ImageDataGenerator(
    rotation_range = 10,
    width_shift_range = 0.1,
    height_shift_range = 0.1,
    zoom_range = 0.2,
    horizontal_flip = True,
    validation_split = 0.1,
)
```

```python
model = tf.keras.models.Sequential([
    tf.keras.layers.Conv2D(32, (3,3), activation='relu', input_shape = [28,28,1]),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Conv2D(64, (3,3), activation = 'relu'),
    tf.keras.layers.MaxPooling2D(2,2),
    tf.keras.layers.Flatten(),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(128, activation = 'relu'),
    tf.keras.layers.Dense(10, activation = 'softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics = ['accuracy'],
)
```

```python
history = model.fit(datagen.flow(x_train, y_train, batch_size=64), epochs=10)
```

```python
# Evaluating the model on the test dataset
loss, acc = model.evaluate(
    x_test, y_test
)

print(f"Accuracy: {acc}, Loss: {loss}")
```

```python
plt.plot(history.history['accuracy'], label='Training Accuracy', marker = '.')
```

```python
plt.plot(history.history['loss'], label = 'Validation Accuracy', marker='.')
```

```python
import cv2
import numpy as np
class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat',
               'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

def predict_cstm_img(img_pth):
  # Loading Image
  img = cv2.imread(img_pth, cv2.IMREAD_GRAYSCALE)
  img = cv2.resize(img, (28, 28))
  img = img/255.0
  img = img.reshape(1, 28, 28, 1)

  # Prediction Part
  predict = model.predict(img)
  predict_class = class_names[np.argmax(predict)]

  plt.imshow(img.reshape(28, 28), cmap = 'gray')
  plt.title(f"Predicted: {predict_class}")
  plt.show()

```
